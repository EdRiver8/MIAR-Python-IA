```python
# Dado una lista de valores, cree un diccionario cuya clave sea el dato de la lista y el valor sea el resto de la división entre 6. 
# No se deben incluir en el diccionario los valores cuyo resto es igual a cero.
def crear_diccionario(valores):
    diccionario = {}
    for valor in valores:
        resto = valor % 6
        if resto != 0:
            diccionario[valor] = resto
    return diccionario

if __name__ == "__main__":
    lista_valores = [10, 12, 15, 18, 20, 24, 30]
    resultado = crear_diccionario(lista_valores)
    print(resultado)  # Salida esperada: {10: 4, 15: 3, 20: 2, 30: 0}
```

Se pide crear un generador que recorra una lista dada y, que en cada llamada, devuelva la media de los valores contiguos y el valor central definido por la posición del índice de dicha lista. Es decir, dada la lista [12, 3, 9, 3, 6], el generador devolverá en la primera iteración la media de 12 y 3 dando como resultado 7.5. En la segunda iteración será 12, 3 y 9, resultando 8. El último valor que devolverá será 3 y 6, resultando 4.5.

```python
# Se pide crear un generador que recorra una lista dada y, que en cada llamada, devuelva la media de los valores contiguos 
# y el valor central definido por la posición del índice de dicha lista. Es decir, dada la lista [12, 3, 9, 3, 6], el 
# generador devolverá en la primera iteración la media de 12 y 3 dando como resultado 7.5. En la segunda iteración 
# será 12, 3 y 9, resultando 8. El último valor que devolverá será 3 y 6, resultando 4.5.

def generador_medias(lista):
    n = len(lista)
    for i in range(n):
        if i == 0:
            yield (lista[i] + lista[i + 1]) / 2
        elif i == n - 1:
            yield (lista[i - 1] + lista[i]) / 2
        else:
            yield (lista[i - 1] + lista[i] + lista[i + 1]) / 3

if __name__ == "__main__":
    lista_ejemplo = [12, 3, 9, 3, 6]
    generador = generador_medias(lista_ejemplo)
    for media in generador:
        print(media)  # Salidas esperadas: 7.5, 8.0, 5.0, 6.0, 4.5
```

```python
ls_num = [12, 3, 9, 3, 6]
gen = generador_medias(ls_num)
ls_res = []
ls_solucion = [7.5, 8.0, 5.0, 6.0, 4.5]
for i in gen:
    ls_res.append(i)
assert (ls_res==ls_solucion), "Solución incorrecta"
```

```python
# Se pide crear una función que reciba un array de NumPy y devuelva otro en el cual estén marcados como True los elementos
#  duplicados (a partir de la segunda ocurrencia), y como False los no repetidos o las primeras ocurrencias de los duplicados.?
import numpy as np
def marcar_duplicados(arr):
    vistos = set()
    resultado = np.zeros(arr.shape, dtype=bool) # Inicializa un array de booleanos con False
    for i in range(len(arr)):
        if arr[i] in vistos:
            resultado[i] = True
        else:
            vistos.add(arr[i])
    return resultado

if __name__ == "__main__":
    array_ejemplo = np.array([1, 2, 3, 2, 4, 1, 5])
    resultado = marcar_duplicados(array_ejemplo)
    print(resultado)  # Salida esperada: [False False False  True False  True False]
```
