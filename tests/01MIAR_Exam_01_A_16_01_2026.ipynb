```python
# - Dado un array de elementos, escribir una función, que reciba dicho array como parámetro de entrada y devuelva el elemento repetido. Usando plain Python, sin ninguna librería.
def find_repeated_element(arr):
    seen = set()
    for element in arr:
        if element in seen:
            return element
        seen.add(element)
    return None

# Test Use Case
test_array = [1, 2, 3, 4, 5, 3, 6, 7]
expected_repeated = 3
output_repeated = find_repeated_element(test_array)
print(f"Expected: {expected_repeated}, Found: {output_repeated}")
```

- Resolución de dos sistemas de ecuaciones
    - $ A : \begin{cases} 8x +  2y = 11 \\ -3y + 10x = 44                         \end{cases}$
    
    - $ B : \begin{cases} -4z + y = 5 \\ 3y + 2z = -2 \\ -x -z + 2y = -1 \end{cases}$
- Verifica por código los resultados

```python
# - Resolución de dos sistemas de ecuaciones
#     - $ A : \begin{cases} 8x +  2y = 11 \\ -3y + 10x = 44                         \end{cases}$
    
#     - $ B : \begin{cases} -4z + y = 5 \\ 3y + 2z = -2 \\ -x -z + 2y = -1 \end{cases}$
# - Verifica por código los resultados

import numpy as np
from scipy.linalg import solve

# System A: 8x + 2y = 11, 10x - 3y = 44
A = np.array([[8, 2], [10, -3]])
b_A = np.array([11, 44])
solution_A = solve(A, b_A)
print("Solution A (x, y):", solution_A)

# System B: -4z + y = 5, 3y + 2z = -2, -x - z + 2y = -1
B = np.array([[0, 1, -4], [0, 3, 2], [-1, 2, -1]])
b_B = np.array([5, -2, -1])
solution_B = solve(B, b_B)
print("Solution B (x, y, z):", solution_B)
```
