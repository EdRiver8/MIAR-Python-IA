```python
import matplotlib.pyplot as plt
import pandas as pd
# sirve para que los gráficos se muestren dentro del cuaderno
%matplotlib inline 
```

```python
numeros = [7, 2, 9, 2, 8]
plt.plot(numeros)
plt.title("Gráfico de Números")
plt.xlabel("Índice")
```

```python
fig = plt.figure(facecolor='lightgray')
ax = plt.axes()
```

```python
# modificar la figura de fondo
plt.figure(facecolor='lightgray')
plt.plot(numeros)
```

```python
plt.subplots(ncols=2, nrows=2) # crear una figura con dos gráficos en dos filas
```

```python
# los axes son los ejes de los gráficos, se pueden modificar individualmente
fig, axes = plt.subplots(ncols=2, nrows=2) # crear una figura con dos gráficos en dos filas y se almacenan en 'axes'
axes[0, 0].plot(numeros, 'r*:')  # gráfico rojo con líneas discontinuas
axes[0, 1].bar([0, 1, 2, 3, 4], numeros)  # gráfico de barras
axes[1, 0].scatter([0, 1, 2, 3, 4], numeros)  # gráfico de dispersión
axes[1, 1].pie(numeros, labels=['A', 'B', 'C', 'D', 'E'])  # gráfico circular
fig.tight_layout()  # ajustar el espaciado entre gráficos
plt.show()
```

```python
ruta_lluvias = ('Lluvias_region_A.csv')
lluvias_df = pd.read_csv(ruta_lluvias)
lluvias_df.head()
```

```python
plt.plot(lluvias_df['region'], lluvias_df['enero'])
plt.title("Lluvias en Enero por Región")
plt.xlabel("Región")
plt.ylabel("Lluvias (mm)")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
ruta_ventas = ('Ventas.csv')
ventas_df = pd.read_csv(ruta_ventas)
ventas_df.head()
```

```python
plt.hist(ventas_df['Producto'], bins=10, color='skyblue', edgecolor='black')
plt.title("Histograma de Monto de Ventas")
plt.xlabel("Monto de Venta")
plt.ylabel("Frecuencia")
plt.grid()
plt.show()
```

```python
# Scatter es un gráfico de dispersión que permite visualizar la relación entre dos variables numéricas.
plt.scatter(ventas_df['Cantidad'], ventas_df['Precio Unitario'])
plt.title("Dispersión entre Cantidad y Monto de Ventas")
plt.xlabel("Cantidad")
plt.ylabel("Monto de Venta")
plt.grid()
```
