## Limpiar los datos para poder usarlos
### que existan errores, falten datos, formatos incorrectos, eliminar duplicados

```python
productos = {"id_producto": [1001, 1002, 1003, 1003],
        "cantidad_vendida": [30, None, 25, 25],
        "precio": [20.5, 15.5, None, 22.5]}
productos
```

```python
productos_df = pd.DataFrame(productos)
productos_df
```

```python
productos_df.head()
```

```python
productos_df.info()
```

```python
productos_df.isnull() # indica un true donde el valor es nulo
```

```python
productos_df.isnull().sum()
```

```python
# conociendo donde estan los valores faltantes, se toma una decision segun el caso
# dependiendo, se pueden eliminar pero se sabe que afecta las estadisticas,
# quizas sea en ocasiones cambiarlo por otro tipo de valores como puede ser la media

# Caso Eliminar:
df_eliminados = productos_df.dropna()
df_eliminados
```

```python
# Caso Cambiar Valores por Otros
valores_nuevos = {'cantidad_vendida': 0, 'precio': productos_df["precio"].mean()}
df_rellenados = productos_df.fillna(valores_nuevos)
df_rellenados
```

```python
# cambiar los tipos de datos al correcto, en este caso pasa de float a entero en cantidad vendida
df_rellenados["cantidad_vendida"] = df_rellenados["cantidad_vendida"].astype(int)
df_rellenados
```

```python
# Verificar si hay duplicados y tomar decisiones con relacion a ello, si dejarlos o eliminarlos o sumarlos
df_unicos = df_rellenados.drop_duplicates() # compara que las filas sean completamente iguales para eliminar
df_unicos
```

```python
# no elimino nada porque no habian filas identicas
df_unicos = df_rellenados.drop_duplicates(subset= ["id_producto"]) # se eliminan las filas que tienen datos duplicados
df_unicos
```

```python
## Concatenar dos o mas DF
personas2 = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Sara'],
    'Edad': [25, 30, 22, 27]
})
personas3 = pd.DataFrame({
    'Nombre': ['Carmela', 'Max', 'Carlos', 'Laura'],
    'Edad': [27, 35, 28, 27]
})
```

```python
personas_unidas = pd.merge(personas2, personas3, on="Nombre")
personas_unidas
```

```python
personas_unidas_inner = pd.merge(personas2, personas3, on="Nombre", how="inner")
personas_unidas_inner
```

```python
personas_concatenado = pd.concat([personas2, personas3])
personas_concatenado
```

```python
personas_concatenado = pd.concat([personas2, personas3], axis=1)
personas_concatenado
```

```python
personas_concatenado = pd.concat([personas2, personas3], ignore_index=True)
personas_concatenado
```

```python
personas_concatenado = pd.concat([personas2, personas3],keys=['personas2', 'personas3'])
personas_concatenado
```
