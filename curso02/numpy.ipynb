## Limpiar Datos con Numpy

```python
array_con_faltante = np.array([[1, 2, np.nan], [4, np.nan, 6], [7, 8, 9]])
array_con_faltante
```

```python
np.isnan(array_con_faltante)
```

```python
# promedio ignorando los NaN
np.nanmean(array_con_faltante)
```

```python
# usar where para reemplazar valores NaN
array_sin_nan = np.where(np.isnan(array_con_faltante), 0, array_con_faltante) # funciona como un if de excelS
array_sin_nan
```

```python
promedio_actualizado = np.mean(array_sin_nan)
promedio_actualizado
```

```python
# Eliminar los datos faltantes
array_sin_nan_eliminado = array_con_faltante[~np.isnan(array_con_faltante)]
array_sin_nan_eliminado
```

```python
# Importar y exportar datos con NumPy
np.savetxt('array_sin_nan.csv', array_sin_nan, delimiter=',')
# array_cargado = np.loadtxt('array_sin_nan.csv', delimiter=',')
array_cargado = np.genfromtxt('array_sin_nan.csv', delimiter=',', filling_values=0, dtype=float) # si hay NaN los rellena con 0
# array_cargado = np.genfromtxt('array_sin_nan.csv', delimiter=',', skip_header=0) # si hay cabecera la salta
array_cargado
```
